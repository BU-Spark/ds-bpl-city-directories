```python
from pathlib import Path
import requests
import pandas as pd
```

This code currently only finds the text and image files for 1916 directory A-O.

Identifier for O-Z: bostonmassachuse19162112samp

```python
# find the different filenames and types attached to 1916 Boston Directory, Vol. 1 (A–O) 
identifier = "bostonmassachuse1916112samp"
data = requests.get(
    f"https://archive.org/metadata/{identifier}"
).json()

for f in data["files"]:
    name = f["name"].lower()
    if any(ext in name for ext in (
        "djvu.txt", "hocr", "chocr",
        "djvu.xml", "jp2", "abbyy"
    )):
        print(f["name"], f.get("size"))
```

Plain text page 162 starts at line 53148

```python
# retrieve plain text file
base_url = f"https://archive.org/download/{identifier}"

output_dir = Path("fa26/data/1916 Boston Directory Vol. 1 (A–O)")
output_dir.mkdir(exist_ok=True)

plain_text_file = "bostonmassachuse1916112samp_djvu.txt"

url = f"{base_url}/{plain_text_file}"
response = requests.get(url, stream=True)
response.raise_for_status()

output_path = output_dir / plain_text_file
with open(output_path, "wb") as f:
    # will download in smaller peieces rather than all at once (1 MB chunks)
    for chunk in response.iter_content(chunk_size=1024 * 1024):
        if chunk:
            f.write(chunk)

    print(f"Downloaded: {output_path}")
```

```python
with open("../data/1916 Boston Directory Vol. 1 (A–O)/bostonmassachuse1916112samp_djvu.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

# line 52,949 in the file is the start of printed page 162
start_line = 53148

page_162_onward = lines[start_line - 1:]

with open("../data/1916 Boston Directory Vol. 1 (A–O)/directory_page162_onward_A-O.txt", "w", encoding="utf-8") as f:
    f.writelines(page_162_onward)

print(f"Kept {len(page_162_onward):,} lines")
```

```python
# retrieve image files
img_file = "bostonmassachuse1916112samp_jp2.zip"

url = f"{base_url}/{img_file}"
response = requests.get(url, stream=True)
response.raise_for_status()

output_path = output_dir / img_file
with open(output_path, "wb") as f:
    for chunk in response.iter_content(chunk_size=1024 * 1024):
        if chunk:
            f.write(chunk)

    print(f"Downloaded: {output_path}")
```

# Create Dataframe of raw text file 

```python
# Load non-blank raw text lines into a DataFrame
with open("../data/1916 Boston Directory Vol. 1 (A–O)/directory_page162_onward_A-O.txt", "r", encoding="utf-8") as file:
    lines = [line.strip() for line in file if line.strip()]

df = pd.DataFrame(lines, columns=["raw_line"])

# Display the first few rows
print(df.head(10))
```

```python
# Recreate cleaned csv of raw lines
df.to_csv('../data/A-O raw lines.csv')
```
